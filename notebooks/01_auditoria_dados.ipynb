# Auditoria Financeira — Pedidos × Pagamentos

## Objetivo

Verificar a consistência financeira dos pedidos, comparando o valor esperado da transação com o valor efetivamente registrado na tabela de pagamentos.

A análise considera como valor esperado do pedido:

**valor dos produtos + valor do frete**

Esse valor é comparado ao total dos pagamentos registrados para o mesmo `order_id`.

---

## 1. Construção do valor total do pedido

Os valores dos itens foram agregados por pedido:

- `total_price`: soma dos preços dos itens;
- `total_freight`: soma dos valores de frete;
- `total_order_value`: valor total esperado do pedido.

O cálculo utilizado foi:

`total_order_value = total_price + total_freight`

---

## 2. Construção do valor total pago

Os registros da tabela `order_payments` foram agrupados por `order_id`, somando todos os pagamentos associados a cada pedido.

Foi criada a variável:

`total_payment_value`

representando o valor total registrado como pago para cada pedido.

---

## 3. Comparação financeira

As duas informações foram relacionadas por `order_id`.

Foi criada a variável:

`payment_difference = total_payment_value - total_order_value`

Interpretação:

- `payment_difference = 0` → valores compatíveis;
- `payment_difference > 0` → pagamento superior ao valor calculado;
- `payment_difference < 0` → pagamento inferior ao valor calculado.

---

## 4. Tratamento de diferenças de ponto flutuante

Durante a comparação foram identificadas pequenas diferenças numéricas próximas de zero, compatíveis com limitações de representação de números decimais em operações computacionais.

Por isso foi adotada uma tolerância de:

`tolerance = 0.01`

Foram consideradas divergências reais apenas as observações em que:

`abs(payment_difference) > 0.01`

---

## 5. Resultado geral

Foram identificadas:

- **381 divergências financeiras reais**;
- **0,39%** dos pedidos analisados apresentaram divergências acima da tolerância.

Portanto, **99,61% dos pedidos analisados não apresentaram divergência financeira superior a R$ 0,01**.

---

## 6. Direção das divergências

Entre as 381 divergências reais:

- **291 (76,38%)** apresentaram pagamento superior ao valor calculado;
- **90 (23,62%)** apresentaram pagamento inferior ao valor calculado.

As divergências positivas foram, portanto, predominantes.

---

## 7. Magnitude das divergências

### Divergências negativas

Foram identificadas 90 ocorrências.

- média: aproximadamente **-R$ 2,22**;
- mediana: aproximadamente **-R$ 0,01**;
- mínimo: **-R$ 51,62**.

A mediana próxima de zero indica que grande parte dessas divergências é pequena, apesar da existência de alguns casos mais extremos.

### Divergências positivas

Foram identificadas 291 ocorrências.

- média: aproximadamente **R$ 10,55**;
- mediana: aproximadamente **R$ 5,55**;
- máximo: **R$ 182,81**.

As divergências positivas apresentam maior magnitude e merecem atenção especial nos casos extremos.

---

## 8. Investigação das divergências positivas

A análise dos meios de pagamento mostrou predominância de cartões de crédito entre as divergências positivas.

Também foram observados pedidos com múltiplos registros de pagamento, embora a maioria das divergências positivas esteja associada a apenas um registro de pagamento.

Entre as 291 divergências positivas:

- **272 pedidos** possuíam um registro de pagamento;
- **19 pedidos** possuíam dois registros de pagamento.

Nos casos analisados, o cartão de crédito foi o principal meio de pagamento associado às divergências positivas.

---

## 9. Relação com parcelamento

Foi analisada a relação entre `payment_difference` e `payment_installments`.

A correlação de Pearson observada foi aproximadamente:

`r = 0.60`

indicando uma **associação positiva moderada** entre número de parcelas e magnitude da divergência positiva.

Além disso, a análise agrupada mostrou tendência de aumento da divergência média conforme aumentava o número de parcelas.

Esse resultado indica que pedidos com maior quantidade de parcelas podem representar um grupo relevante para investigação.

Entretanto, **correlação não implica causalidade**. Portanto, não é possível concluir que o parcelamento seja a causa das divergências apenas com essa análise.

---

## 10. Casos extremos

Entre os maiores valores de divergência positiva foram encontrados pedidos com diferenças superiores a R$ 100.

O maior caso identificado apresentou:

- valor dos produtos: **R$ 1.299,00**;
- frete: **R$ 104,66**;
- valor esperado: **R$ 1.403,66**;
- valor registrado como pago: **R$ 1.586,47**;
- divergência: **R$ 182,81**.

A investigação individual desses casos mostrou que as maiores divergências não estavam necessariamente relacionadas à existência de múltiplos pagamentos.

Também foram observados pedidos com quantidade elevada de parcelas.

## 11. Conclusão da auditoria

A auditoria financeira indica que os registros de pedidos e pagamentos apresentam **alta consistência geral**, uma vez que apenas **0,39%** dos pedidos analisados apresentaram divergência superior à tolerância estabelecida de R$ 0,01.

Apesar da baixa proporção, as divergências não devem ser ignoradas, principalmente devido à existência de casos com valores elevados.

As principais evidências encontradas foram:

- predominância de divergências positivas;
- maior magnitude média das divergências positivas;
- predominância de cartão de crédito entre essas ocorrências;
- associação moderada entre número de parcelas e magnitude da divergência;
- existência de poucos casos extremos que merecem investigação individual.
